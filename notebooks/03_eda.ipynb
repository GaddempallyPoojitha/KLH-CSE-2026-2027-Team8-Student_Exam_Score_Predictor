```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

dataset_path = r"C:\Student-Exam-Score-Predictor\dataset\student_data_cleaned.csv"

df = pd.read_csv(dataset_path)

print("Cleaned dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Exam_Score"], bins=20, kde=True)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Hours_Studied",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Daily Study Hours")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam_Score"
)

plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Previous_Scores",
    y="Exam_Score"
)

plt.title("Previous Scores vs Exam Score")
plt.xlabel("Previous Score")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Exam_Score"
)

plt.title("Gender vs Exam Score")
plt.xlabel("Gender")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Family_Income",
    y="Exam_Score"
)

plt.title("Family Income vs Exam Score")
plt.xlabel("Family Income")
plt.ylabel("Exam Score")
plt.show()
```

```python
attendance_groups = pd.cut(
    df["Attendance"],
    bins=[0, 60, 75, 90, 100],
    labels=["Below 60", "60-75", "75-90", "90-100"]
)

attendance_analysis = df.groupby(
    attendance_groups,
    observed=False
)["Exam_Score"].mean()

print(attendance_analysis)

plt.figure(figsize=(8, 5))

attendance_analysis.plot(kind="bar")

plt.title("Average Exam Score by Attendance Group")
plt.xlabel("Attendance Range")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=0)
plt.show()
```

```python
numerical_df = df.select_dtypes(include=["int64", "float64"])

correlation_matrix = numerical_df.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```
