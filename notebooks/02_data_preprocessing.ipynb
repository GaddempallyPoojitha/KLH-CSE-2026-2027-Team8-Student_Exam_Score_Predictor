```python
import pandas as pd
import numpy as np

dataset_path = r"C:\Student-Exam-Score-Predictor\dataset\StudentPerformanceFactors.csv"

df = pd.read_csv(dataset_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Missing values before preprocessing:")
print(df.isnull().sum())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

print("Numerical columns:")
print(list(numerical_columns))

print("\nMinimum values:")
print(df[numerical_columns].min())

print("\nMaximum values:")
print(df[numerical_columns].max())
```

```python
missing_columns = [
    "Teacher_Quality",
    "Parental_Education_Level",
    "Distance_from_Home"
]

for column in missing_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing categorical values filled successfully.")
```

```python
print("Missing values after preprocessing:")
print(df.isnull().sum())
```

```python
print("Duplicate rows after preprocessing:", df.duplicated().sum())
```

```python
cleaned_path = r"C:\Student-Exam-Score-Predictor\dataset\student_data_cleaned.csv"

df.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print(cleaned_path)
```
